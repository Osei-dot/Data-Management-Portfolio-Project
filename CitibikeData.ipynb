```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import glob
```

```python
files = glob.glob('JC-2016*-citibike-tripdata.csv')
```

```python
print(files)
```

```python
df_list = [pd.read_csv(file) for file in files]
```

```python
df = pd.concat(df_list)
```

```python
split_repdate = df['Start Time'].str.split('-', expand=True)[0]
```

```python
split_repdate
```

```python
df['year_of_ride'] = pd.to_numeric(split_repdate)
```

```python
df.head()
```

```python
df['Start Time']=pd.to_datetime(df['Start Time'])
df['Stop Time'] = pd.to_datetime(df['Stop Time'])
```

```python
df['Month'] = df['Start Time'].dt.strftime('%m')
```

```python
df['Gender'].value_counts()
```

```python
df['Gender'].replace(1, 'Male', inplace = True)
df['Gender'].replace(2, 'Female', inplace = True)
df['Gender'].replace(0, 'Unknown', inplace = True)
```

```python
df['Gender'] =df['Gender'].astype(str)
```
